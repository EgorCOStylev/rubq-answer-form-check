```python
import importlib, subprocess, sys
PKG = {'fastchat': 'fschat', 'hydra': 'hydra-core', 'bs4': 'beautifulsoup4', 'bert_score': 'bert-score', 'sklearn': 'scikit-learn'}
MODULES = ['lm_polygraph.model_adapters', 'lm_polygraph.utils.generation_parameters', 'lm_polygraph.stat_calculators.sample']
for _ in range(30):
    try:
        for m in MODULES:
            importlib.import_module(m)
        print('lm_polygraph import ok')
        break
    except ModuleNotFoundError as e:
        name = e.name.split('.')[0]
        pkg = PKG.get(name, name)
        print('installing', pkg)
        extra = ['--no-deps'] if pkg == 'fschat' else []
        subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', *extra, pkg], check=True)
```
